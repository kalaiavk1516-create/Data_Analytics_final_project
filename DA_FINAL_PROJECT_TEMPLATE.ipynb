4. Outlier Detection and Treatment

Outliers were checked using boxplots.

Python
import matplotlib.pyplot as plt

plt.boxplot(df['AMOUNT'])
plt.title("Deposit Amount Outliers")
plt.show()

Using IQR Method:

Python
Q1 = df['AMOUNT'].quantile(0.25)
Q3 = df['AMOUNT'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[(df['AMOUNT'] >= lower_bound) &
(df['AMOUNT'] <= upper_bound)]

```python
#Year-wise Total Deposit Amount (chart)

year_deposits = df.groupby("year")["deposit_amount"].sum()
plt.figure(figsize=(10,5))

plt.plot(
    year_deposits.index,
    year_deposits.values,
    marker="o",
)

plt.title("Total Deposit Amount by Year")
plt.xlabel("Year")
plt.ylabel("Total Deposit Amount")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#Population Group-wise Deposits

# Fix: Change 'data' to 'df'
group_deposits = df.groupby("population_group")["deposit_amount"].sum().sort_values(ascending=False)

group_deposits.plot(kind="bar", figsize=(9,5))
plt.title("Total Deposit Amount by Population Group")
plt.xlabel("Population Group")
plt.ylabel("Total Deposit Amount")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

```python
#Top 10 States by Deposits

top_states = df.groupby("state_name")["deposit_amount"].sum().nlargest(10).sort_values()

top_states.plot(kind="barh", figsize=(10,6))
plt.title("Top 10 States by Deposits")
plt.xlabel("Total Deposit Amount")
plt.ylabel("State")
plt.tight_layout()
plt.show()
```

```python
#Region-wise Total Deposits

region_deposits = df.groupby("region")["deposit_amount"].sum().sort_values(ascending=False)

region_deposits.plot(kind="pie",autopct="%1.1f%%", figsize=(9,5))
plt.title("Total Deposits by Region")
plt.xlabel("Region")
plt.ylabel("Total deposit amount")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Population Group-wise Number of Accounts


group_accounts = (
    df.groupby("population_group")["no_of_accounts"]
    .sum()
    .sort_values(ascending=False)
)

group_accounts.plot(kind="bar", figsize=(9,5))
plt.title("Total Number of Accounts by Population Group")
plt.xlabel("Population Group")
plt.ylabel("Total Number of Accounts")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

```python
#Region-wise Number of Offices


region_offices = (
    df.groupby("region")["no_of_offices"]
    .sum()
    .sort_values(ascending=False)
)

region_offices.plot(kind="pie", autopct="%1.1f%%", figsize=(9,5))
plt.title("Total Number of Offices by Region")
plt.xlabel("Region")
plt.ylabel("Total Number of Offices")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
#Deposit Amount Distribution


plt.figure(figsize=(10,5))

plt.hist(df["deposit_amount"].dropna(), bins=20)

plt.title("Distribution of Deposit Amounts")
plt.xlabel("Deposit Amount")
plt.ylabel("number of records")
plt.tight_layout()
plt.show()
```
